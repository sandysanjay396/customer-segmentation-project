```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Screen-la irukura sample_data/california_housing_train.csv read pandrom
df = pd.read_csv('sample_data/california_housing_train.csv')

# 2. Key features select pandrom
X = df[['median_income', 'median_house_value']]

# 3. Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. K-Means Model
kmeans = KMeans(n_clusters=4, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

# 5. Graph Plotting
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df.sample(500),
    x='median_income',
    y='median_house_value',
    hue='Cluster',
    palette='Set1'
)
plt.title('Customer / Housing Segmentation Analysis')
plt.grid(True)
plt.show()

# Summary
print(df.groupby('Cluster')[['median_income', 'median_house_value']].mean())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Sample Data (Unkitta dataset illana idhe podhum)
np.random.seed(42)
data = {
    'CustomerID': range(1, 201),
    'Age': np.random.randint(18, 70, 200),
    'Annual_Income': np.random.randint(15, 140, 200),
    'Spending_Score': np.random.randint(1, 100, 200)
}
df = pd.DataFrame(data)

# 2. Feature Scaling
X = df[['Age', 'Annual_Income', 'Spending_Score']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. K-Means Model Training
kmeans = KMeans(n_clusters=4, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

# 4. Visualization
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='Annual_Income',
    y='Spending_Score',
    hue='Cluster',
    palette='viridis',
    s=80
)
plt.title('Customer Segments (Income vs Spending Score)')
plt.grid(True)
plt.savefig('cluster_plot.png') # Graph image save aagum
plt.show()

# 5. Insights Summary
print(df.groupby('Cluster')[['Age', 'Annual_Income', 'Spending_Score']].mean())
```
